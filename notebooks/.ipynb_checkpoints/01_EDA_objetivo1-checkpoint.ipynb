**11. Revisión de columnas numéricas guardadas como texto**

Primero miramos en las columnas que deberían ser numéricas 

```python
for columna in ["precio_unidad", "unidades_vendidas"]:
    valores = df_eda[columna].dropna().astype(str).str.strip()
    conversion = pd.to_numeric(
        valores.str.replace(",", ".", regex=False),
        errors="coerce"
    )

    print(f"\n{columna}:")
    display(valores[conversion.isna()].value_counts().head(20))
```

```python
columnas_revisar = [
    "id_plato", "categoria", "valoracion", "dificultad_elaboracion"
]

df_eda[columnas_revisar].isna().sum(axis=1).value_counts().sort_index()
```

Ya encontramos los casos que no se podían interpretar como números:
- precio_unidad: "12,5 €" aparece 1.358 veces.
- unidades_vendidas: "muchos" aparece 1.351 veces.
- En 737 filas faltan a la vez id_plato, categoria, valoracion y dificultad_elaboracion;

Seguimos conservando el dataset original. 

```python
# Convertimos el precio a número: vamos a quitar el signo de la moneda (€) y cambiamos la coma por punto.
precio_limpio = (
    df_eda["precio_unidad"]
    .astype("string")
    .str.replace("€", "", regex=False)
    .str.replace(",", ".", regex=False)
    .str.strip()
)
df_eda["precio_unidad_num"] = pd.to_numeric(
    precio_limpio,
    errors="coerce"
)
# Vamos a vaciar donde el valor es "muchos" 
unidades_limpias = (
    df_eda["unidades_vendidas"]
    .astype("string")
    .str.strip()
    .str.lower()
    .replace({"muchos": pd.NA})
)
df_eda["unidades_vendidas_num"] = pd.to_numeric(
    unidades_limpias,
    errors="coerce"
)
```

**15b. ¿El nombre del plato tiene relación con la categoría?**

Al mirar los datos vemos platos con nombre de ensalada o de croquetas con la categoría `postre`. Comprobamos qué relación hay de verdad entre `id_plato`, `nombre_plato_aux` y `categoria`.

```python
# Trabajamos solo con filas que tienen id y categoría
d = df_eda.dropna(subset=["id_plato", "categoria"]).copy()

# Mismo plato = mismo número (quitamos el prefijo PL-); el nombre base es el texto sin el número final
d["id_num"] = d["id_plato"].astype(str).str.replace("PL-", "", regex=False)
d["nombre_base"] = d["nombre_plato_aux"].str.replace(r"\s*\d+$", "", regex=True).str.strip()

print("Categorías distintas por plato (máximo):", d.groupby("id_num")["categoria"].nunique().max())
print("Nombres base distintos por plato (mín., máx.):",
      d.groupby("id_num")["nombre_base"].nunique().agg(["min", "max"]).tolist())

# Reparto de categorías para cada nombre base (cada fila suma 1)
display(pd.crosstab(d["nombre_base"], d["categoria"], normalize="index").round(2))

# Ejemplo: un mismo plato con todos los nombres que recibe
display(d[d["id_num"] == "46"][["nombre_plato_aux", "categoria"]].value_counts().head(10).to_frame("filas"))
```

**Qué pasa:** cada `id_plato` tiene una única categoría, pero su `nombre_plato_aux` cambia de una fila a otra (10 nombres distintos por plato). Además, cada nombre se reparte igual entre las cuatro categorías, así que el nombre no dice nada de la categoría. Por eso el plato 46 aparece como "Ensalada 46" y como "Croquetas 46" y los dos son `postre`.

**Decisión:**
- El identificador del plato es `id_plato`. `nombre_plato_aux` no es fiable y se elimina (como ya hacíamos).
- `categoria` se queda, porque es un atributo estable de cada plato (y el profesor aconseja conservarla), pero **no hay que mezclarla con el nombre** en tablas ni gráficos. En informes llamaremos a los platos "Plato 46", no "Ensalada 46".
- En el informe hay que decir que la categoría de estos datos sintéticos se asignó al plato sin relación con su nombre.

**18. Limpieza de datos**

Creamos `df_limpio` a partir de `df_eda`. El criterio es **no rellenar con la media ni con la mediana** las columnas con las que calcularemos la popularidad y el margen (`unidades_vendidas`, `clientes_dia`, `precio_unidad` y `coste_unidad`). Un valor inventado se mezcla con los reales y hace que esas métricas dejen de cuadrar. Las filas sin esos datos se eliminan más abajo, y la popularidad y el margen se recalculan.

Antes de eso corregimos dos cosas:
- `id_plato`: el mismo plato aparece como `19` y como `PL-19`. Quitamos el prefijo para tener un único id por plato.
- `clientes_dia = 0`: con 0 clientes no se puede calcular la popularidad (división por cero), así que lo tratamos como vacío.

```python
import numpy as np

df_limpio = df_eda.copy()

# 1) id_plato: quitamos el prefijo "PL-" para que 19 y PL-19 sean el mismo plato
df_limpio["id_plato"] = (
    df_limpio["id_plato"].astype("string").str.replace("PL-", "", regex=False).str.strip()
)
print("Platos distintos:", df_limpio["id_plato"].nunique())

# 2) clientes_dia = 0 pasa a vacío (no se puede dividir entre 0)
df_limpio["clientes_dia"] = df_limpio["clientes_dia"].astype(float)
n_ceros = int((df_limpio["clientes_dia"] == 0).sum())
df_limpio["clientes_dia"] = df_limpio["clientes_dia"].replace(0, np.nan)
print("Filas con clientes_dia = 0 pasadas a vacío:", n_ceros)
```

Revisamos qué vacíos tenemos ahora y que valoración y dificultad respetan los rangos que hemos establecido.

```python
columnas_revisar = [
    "id_plato",
    "categoria",
    "precio_unidad_num",
    "coste_unidad",
    "unidades_vendidas_num",
    "clientes_dia",
    "valoracion",
    "dificultad_elaboracion",
]

print("Ausentes por columna:")
display(df_limpio[columnas_revisar].isna().sum())

print("Dificultad fuera de 1–10:",
      ((df_limpio["dificultad_elaboracion"] < 1) |
       (df_limpio["dificultad_elaboracion"] > 10)).sum())

print("Valoración fuera de 1–5:",
      ((df_limpio["valoracion"] < 1) |
       (df_limpio["valoracion"] > 5)).sum())
```

Las columnas originales de texto, como precio_unidad y unidades_vendidas, pueden seguir mostrando ausentes porque conservamos los datos originales; las columnas numéricas derivadas (precio_unidad_num y unidades_vendidas_num) son las que hemos usado.

Los vacíos que hay son:
- 761 fechas en fecha, fecha_dt y fecha_formateada.
- 737 filas sin id_plato, que además no tienen categoría, valoración ni dificultad.
- Filas sin unidades vendidas, sin clientes, sin precio o sin coste.

Las 737 filas sin id (con su categoría) se eliminan más abajo, así que no hace falta inventar una categoría "desconocida".

En las fechas vacías, añadiremos la última fecha válida de la fila anterior. Vamos a conservar la columna original de fecha y solo actualizamos las columnas derivadas

```python
ausentes_antes = df_limpio["fecha_dt"].isna().sum()

df_limpio["fecha_dt"] = df_limpio["fecha_dt"].ffill()
df_limpio["fecha_formateada"] = df_limpio["fecha_dt"].dt.strftime("%d-%m-%Y")

print(f"Fechas ausentes antes: {ausentes_antes}")
print(f"Fechas ausentes después: {df_limpio['fecha_dt'].isna().sum()}")
```

**Eliminamos las filas que no se pueden usar.** Quitamos las filas con vacío en `id_plato`, `unidades_vendidas`, `clientes_dia` (incluidos los 0), `precio_unidad` o `coste_unidad`. A una misma fila le pueden afectar varios motivos, así que la suma de la tabla es mayor que el total eliminado. Las 737 filas sin id son las mismas que no tienen categoría, valoración ni dificultad.

```python
# Motivos por los que una fila no se puede usar
motivos = {
    "sin id_plato": df_limpio["id_plato"].isna(),
    "sin unidades vendidas": df_limpio["unidades_vendidas_num"].isna(),
    "sin clientes (o con 0)": df_limpio["clientes_dia"].isna(),
    "sin precio": df_limpio["precio_unidad_num"].isna(),
    "sin coste": df_limpio["coste_unidad"].isna(),
}
display(pd.DataFrame({"filas": {k: int(v.sum()) for k, v in motivos.items()}}))

descartar = pd.concat(motivos, axis=1).any(axis=1)

filas_antes = len(df_limpio)
df_final = df_limpio[~descartar].copy()
df_final["id_plato"] = df_final["id_plato"].astype(int)     # ya no hay vacíos, pasamos el id a entero

print(f"Filas antes: {filas_antes:,}")
print(f"Filas eliminadas: {filas_antes - len(df_final):,} ({(filas_antes - len(df_final)) / filas_antes:.1%})")
print(f"Filas finales: {len(df_final):,}")
print("Platos distintos:", df_final["id_plato"].nunique())
print("Vacíos en categoría, valoración y dificultad:",
      int(df_final[["categoria", "valoracion", "dificultad_elaboracion"]].isna().sum().sum()))
```

```python
display(df_final.head(10).T)
```

Eliminamos las siguientes columnas porque no nos dan información necesaria para el proyecto:

- fecha: es la fecha original, nos quedamos con la formateada
- nombre_plato_aux: el nombre cambia de una fila a otra para el mismo plato, no es fiable (ver sección 15b)
- moneda: solo tiene un valor
- fecha_dt: es una columna auxiliar para el EDA

**Conservamos** `franja`, `ciudad` y `categoria` siguiendo la recomendación del profesor: no se ven diferencias grandes entre sus valores, pero sirven para describir y entender los grupos de platos más adelante.

```python
# Comprobamos que moneda solo tiene un valor antes de quitarla
print("Valores de moneda:", df_final["moneda"].unique().tolist())

columnas_eliminar = [
    "fecha",
    "nombre_plato_aux",
    "moneda",
    "fecha_dt"
]

df_final = df_final.drop(columns=columnas_eliminar)

print("Dimensiones:", df_final.shape)
print("Columnas restantes:")
df_final.columns.tolist()
```

```python
display(df_final.head(10).T)
```

**Margen y popularidad.** Ahora que no hay valores inventados en precio, coste, unidades ni clientes, recalculamos las dos columnas derivadas y las comparamos con las originales:
- margen_unitario = precio − coste
- popularidad = unidades vendidas / clientes del día

```python
# Margen: precio - coste, comparado con el margen que venía en los datos
margen_nuevo = (df_final["precio_unidad_num"] - df_final["coste_unidad"]).round(2)
dif_margen = (margen_nuevo - df_final["margen_unitario"]).abs()
print("Margen original vacío:", int(df_final["margen_unitario"].isna().sum()))
print("Filas donde el margen original no coincide con precio - coste:", int((dif_margen > 0.011).sum()))

# Popularidad: unidades / clientes, comparada con la que venía en los datos
pop_nueva = (df_final["unidades_vendidas_num"] / df_final["clientes_dia"]).round(4)
dif_pop = (pop_nueva - df_final["popularidad"]).abs()
print("Popularidad original vacía:", int(df_final["popularidad"].isna().sum()))
print("Filas donde la popularidad original no coincide con unidades/clientes:", int((dif_pop > 0.001).sum()))

# Nos quedamos con los valores recalculados (coinciden con los originales y no tienen vacíos)
df_final["margen_unitario"] = margen_nuevo
df_final["popularidad"] = pop_nueva
```

**Columnas finales.** Quitamos las versiones de texto de `precio_unidad` y `unidades_vendidas` y renombramos las numéricas:
- precio_unidad_num → precio_unidad
- unidades_vendidas_num → unidades_vendidas
- fecha_formateada → fecha

Así no se crean columnas duplicadas. También fijamos los tipos y ordenamos las columnas.

```python
# 1) Quitamos las columnas de texto originales
df_final = df_final.drop(columns=["precio_unidad", "unidades_vendidas"])

# 2) Renombramos las numéricas
df_final = df_final.rename(columns={
    "precio_unidad_num": "precio_unidad",
    "unidades_vendidas_num": "unidades_vendidas",
    "fecha_formateada": "fecha",
})

# 3) Tipos: recuentos como enteros, precio como decimal normal
df_final = df_final.astype({
    "id_plato": "int64",
    "unidades_vendidas": "int64",
    "clientes_dia": "int64",
    "dificultad_elaboracion": "int64",
    "precio_unidad": "float64",
    "popularidad": "float64",
    "margen_unitario": "float64",
})

# 4) Orden de las columnas
orden = [
    "id_plato", "fecha", "franja", "ciudad", "categoria",
    "precio_unidad", "coste_unidad", "unidades_vendidas", "clientes_dia",
    "popularidad", "margen_unitario", "valoracion", "dificultad_elaboracion",
]
df_final = df_final[orden]

df_final.columns.tolist()
```
